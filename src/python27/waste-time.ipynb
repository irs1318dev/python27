# Top 10 Ways to Fail to Learn to Program


## #1. Don't establish a routine for learning to code.
> #### Make sure at least two weeks elapse between sessions, so you can forget the old material and have to start all over.

## #2. Avoid reading documentation like the plague.
> #### Assume comprehension is overrated.

## #3. If the code works, just move on.
> #### Don't ask yourself why it worked this time and didn't work last time.

## #4. Use copy and paste as much as you can.
* #### Definitely paste code from the Internet that you don't understand into your application.
* #### Don't practice the details of syntax by typing out code by hand.

## #5. Minimize mental effort.
> #### Immediately turn to AI if a solution is not obvious.

## #6. Don't ever look at anyone else's code.
> #### We don't care how other people approached a problem.

## #7. Don't ever show your code to anyone else.
> #### We hate suggestions that make our code better!

## #8. Assume you are on your own if you get stuck.
> #### Don't talk about concepts or exercises with other students or mentors. 

## #9. Don't try anything that's not mentioned in this course.
> #### Assume your computer will burst into flames!
