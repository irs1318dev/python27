# Comments, Defining Variables and Math
* Use the equals sign (=) to define variables.

```python
# Lines that start with hashes are comments.
# The Python interpreter ignores comments.

# Define a couple variables
a = 3
b = 4

# The two asterisks (**) indicate exponents.
#   So we're squaring both a and b.
c = a**2 + b**2

# What is c?
#   Type it on the last line of the cell and the notebook
#   will display its value.
c
```

```python
print("18 - 13", "=", 18 - 13)
print("10 times 100 =", 10 * 100)
print("64 divided by 4 =", 64 / 4)
print("The remainder of 10 divided by 3 is", 10 % 3)
print("Integer Division: 5 // 2 =", 5//2)
print("Negative numbers: 5 * -3 =", 5 * -3)
```

## Variables
### Storing Stuff with Variables
In computer programming, we often need to save the results of calculations and use them later. And by often, I mean *all the time*. For this we use *variables*. Variables are named values. In Python we create a variable by typing the name of the variable followed by an `=` sign, followed by the value we want the variable to represent.

```python
# Using Variables
radius = 2     # Assignment statement
pi = 3.14      # Assignment statement
height = 5     # Assignment statement

# This line both reads variables and assigns the results of a mathematical expression to
#   a new variable
cylinder_volume = pi * radius**2 * height

# Placing the variable on the final line of the code cell causes Jupyter to
#   display it's value.
cylinder_volume
```

There are a few things to keep in mind regarding variables if you don't want errors.

```python
# Trying to access a variable before it's defined causes an error.
var1 = 5 + undefined_var
```

```python
# Numbers on the left side of the `=` sign cause an error
5 = height
```

## Input Function Always Returns a String!
Math doesn't work on strings.

```python
age = input("How old are you?  ")
print("Next year you will be", age + 1)
```

### int() Function
Use the int() function to convert the string "1" to the integer 1.

```python
age = int(input("How old are you?  "))
print("Next year you will be", age + 1)
```

## float() Function
Use `float()` if you need a floating point number.

```python
import math  # Don't worry about the import statement yet!
pi = float(input("Enter pi to as many digits as you can!  "))
print("You are only off by", pi - math.pi)
```

# If Statement

```python
# Our first `if` statement.
date = int(input("How many days are in April?\n"))
if date == 30:
    print("That's right!")
elif date == 31:
    print("Wrong, but an understandable mistake.")
else:
    print("Not even close!")
```

the "\n" in the input string is a special code that tells Python to print a newline character.

## Comparison Operators

```python
print("--- Equals ---")
print("5 == 5:", 5 == 5)
print("10 == 5:", 10 == 5)

print("\n--- Not Equals ---")
print("5 != 5:", 5 != 5)
print("10 != 5:", 10 != 5)

print("\n--- Greater Than ---")
print("5 > 5:", 5 > 5)
print("10 > 5:", 10 > 5)

print("\n--- Greater Than Or Equal To ---")
print("5 >= 5:", 5 >= 5)
print("10 >= 5:", 10 >= 5)
```

# Boolean Operators `not`, `and`, and `or`

```python
# `not` keyword changes a True value to False and vise versa.
answer = int(input("At what temperature does water boil at sea level (F)?\n"))
if not answer == 100:
    print("Nope!")
else:
    print("Correct!")
```

```python
print("True and False:", True and False)
print("True or False:", True or False)
```

```python
3 > 2 or 100 < 10
```

```python
3 > 2 and 100 < 10
```

```python
3 > 2 and not 100 < 10
```

```python
# Append a number to the end of a list.
my_list.append(10)
my_list
```

Zero-based Indexing!

# While Loops

```python
# Coundown!
count = 10                           # line 2
while count >= 0:                    # line 3
    print(str(count) + "!")          # line 4
    count -= 1                       # line 5
print("Liftoff, we have liftoff!")   # line 6
```
