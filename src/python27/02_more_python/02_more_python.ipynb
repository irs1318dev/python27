## Functions
* #### Functions are the basic building blocks of computer programming
* #### They sometimes have inputs
    * The inputs are usually called *arguments*
* #### They sometimes have an output
    * The output is called *return values*
* #### Sometimes they don't have either!

### We've Already Used Functions!
* #### `sum` tasks a list of numbers and adds them together

````python
total = sum([12, 37, 10])
total
````

* #### Argument: The list `[12, 37, 10]`
* #### Return Value: The integer 59, which is assigned to a variable named `total`.


### The `int` and `float` Functions

````python
my_int = int("1318")
my_int
````

````python
my_float = float("3.14159")
my_float
````

* #### Arguments go inside parentheses!
* #### Use a variable name and the assignment operator `=` to get a return value

### The `print` Function
* #### Takes multiple arguments
* #### Separate arguments with commas

````python
print("Issaquah", "Robotics", "Society")
````

### `print` does not have a return value!

````python
my_val = print("Issaquah", "Robotics", "Society")
````

````python
my_val  # Nothing will be displayed
````

The print method displays output on a screen, but it doesn't have a return value so you can't save anything to a variable.

When a function makes something happen, like displaying stuff on the screen, or deleting a file, we call that a *secondary effect*. The print() function doesn't have a return value, but it does have a *secondary effect*.

You could write a function that has neither a return value nor a secondary effect, but it would not be very useful.

### The `None` Type
* #### We told you that the `print` function doesn't have a return value.
* #### That's not entirely correct

````python
print_return_value = print("I'm printing something!")
type(print_return_value)
````

#### All Python functions return something.
#### But sometimes the something is the `None` object.

````python
# Assign None to a variable
none_var = None
if none_var is None:
    print("none_var contains nothing.")
else:
    print("none_var contains something.")
````

### Boolean Type
#### A Boolean is a value that can either be `True` or `False`.

````python
my_bool = 4 == 5

# Our boolean variable is False
print(my_bool)
# And it has data type bool
print(type(my_bool))   # By the way, checkout how we're nesting functions here!
````

````python
# Setting boolean variables
# Always capitalize the first letter and only the first letter of
#   True and False
true_bool = True
false_bool = False
print(true_bool, false_bool)
````

## Strings
* #### Use single or double quotes to make strings.
* #### (But pick one and use it consistently!)

````python
double_quote = "A problem well stated is a problem half solved."
single_quote = 'A problem well stated is a problem half solved.'
double_quote == single_quote
````

### Multiline Strings
For strings with multiple lines, start and end with 3 quotation marks.

````python
multiline_string = """
This boat that we built is just fine --
And don't try to tell us it's not.
The sides and the back are divine --
It's the bottom I guess we forgot.
"""
# by Shel Silverstein
print(multiline_string)
````

### Getting the length of a string:

````python
# Yay! A new function!
print(multiline_string)
print()
print("Poem length:", len(multiline_string), "characters")
````

### Getting parts of a string

````python
# First 10 characters of poem:
multiline_string[:10]
````

* #### `\n` is a special code that stands for a *newline* character
    * It occurs whenever one line ends and new one begins.
    * Even though it uses two characters when displayed, it only counts as one
* #### Note that the space also counts as a character.

### Get the first character

````python
multiline_string[0]
````

#### Very Important: The first character is at position 0, not 1!

### Next 10 characters

````python
# Print characters at position 10, 11, 12, 12, 14, 15, 16, 17,
# 18, and 19, but don't print the character at position 20
multiline_string[10:20]
````

````python
# character at position 20
multiline_string[20]
````

#### Final 10 characters

````python
# Negative values count from the END of the string.
multiline_string[-10:]
````

#### Skipping Characters

````python
# Print every other character!
print(multiline_string[::2])
````

````python
# Print every 4th character starting with the second!
print(multiline_string[1::4])
````

## For Loops

````python
name = "Jebidiah"
for letter in name:
    print(letter)
````

### For Loop with Numbers

````python
# Count to 5 like the Count!
for num in range(1, 5):
    print(num, "Ah, Ah, Ahhhh")
````

#### The `range` function helps us loop over numbers.
#### But we didn't make it to 5!
`range(1, 5)` starts at 1 and stops right before 5.

````python
# Count to 5 for real this time
for num in range(1, 6):
    print(num, "Ah, Ah, Ahhhh")
````

````python
# Count by 3s
for num in range(3, 20, 3):
    print(num)
````
